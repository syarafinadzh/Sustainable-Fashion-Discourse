```python

!pip install bertopic sentence-transformers openpyxl umap-learn hdbscan --quiet

import pandas as pd
import numpy as np
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import CountVectorizer
from umap import UMAP
import hdbscan
```

```python

file_path = "/content/data clean social stage 1 .xlsx"

df = pd.read_excel(file_path)

if "text" in df.columns:
    docs = df["text"].dropna().astype(str).tolist()
else:
    docs = df.iloc[:,0].dropna().astype(str).tolist()

print("Total documents:", len(docs))
```

```python

vectorizer_model = CountVectorizer(
    stop_words="english",
    ngram_range=(1,2),
    min_df=5
)
```

```python

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python

umap_model = UMAP(
    n_neighbors=15,
    n_components=5,
    min_dist=0.0,
    metric='cosine',
    random_state=42
)

hdbscan_model = hdbscan.HDBSCAN(
    min_cluster_size=20,
    min_samples=10,
    metric='euclidean',
    cluster_selection_method='eom',
    prediction_data=True
)
```

```python

social_seed_topics = [
    ["labor", "labour", "workers", "factory", "garment workers"],
    ["exploitation", "forced labor", "child labour", "modern slavery"],
    ["wages", "fair wages", "minimum wage", "income inequality"],
    ["supply chain", "outsourcing", "developing countries", "global south"],
    ["ethical production", "human rights", "worker rights", "fair trade"]
]
```

```python

topic_model = BERTopic(
    embedding_model=embedding_model,
    vectorizer_model=vectorizer_model,
    umap_model=umap_model,
    hdbscan_model=hdbscan_model,
    seed_topic_list=social_seed_topics,
    calculate_probabilities=True,
    verbose=True
)

topics, probs = topic_model.fit_transform(docs)
```

```python

topic_info = topic_model.get_topic_info()
topic_info
```

```python

valid_topics = topic_info[topic_info["Topic"] != -1]["Topic"].tolist()

fig = topic_model.visualize_barchart(
    topics=valid_topics,
    n_words=10,
    title="Social Topic Words Score"
)

fig.write_html("topic_barchart_Social.html")
print("✓ Saved: topic_barchart_Social.html")
fig.show()
```

```python
import pandas as pd
import numpy as np

# ============================
# BUILD SUMMARY TABLE
# ============================

# Total dokumen
total_docs = len(topics)

# Hitung jumlah dokumen per topic
topic_counts = pd.Series(topics).value_counts().sort_index()

# Hitung total tanpa noise (-1)
total_wo_noise = total_docs - topic_counts.get(-1, 0)

rows = []

for topic_id, n_doc in topic_counts.items():

    # Persentase dari total dokumen
    percent = (n_doc / total_docs) * 100

    # Ambil keywords top 5
    if topic_id != -1:
        words = topic_model.get_topic(topic_id)
        keywords = ", ".join([w[0] for w in words[:5]])

        # Persentase tanpa noise
        percent_wo_noise = (n_doc / total_wo_noise) * 100 if total_wo_noise > 0 else 0
    else:
        keywords = "Noise / Outliers"
        percent_wo_noise = None

    rows.append({
        "Topic": topic_id,
        "n_doc": n_doc,
        "%": round(percent, 2),
        "keywords": keywords,
        "%_wo_noise": round(percent_wo_noise, 2) if percent_wo_noise is not None else None
    })

summary_table = pd.DataFrame(rows).sort_values("Topic").reset_index(drop=True)

summary_table
```

```python
# =========================================
# PICK & RELABEL SOCIAL CORE TOPICS
# Target: old topics [0,4,23,30] -> new topics [0,1,2,3]
# =========================================

selected_topics = [0, 4, 23, 30]
topic_mapping  = {0:0, 4:1, 23:2, 30:3}

# 1) Filter documents + old topic labels
filtered_docs = []
filtered_topics_old = []

for doc, t in zip(docs, topics):
    if t in selected_topics:
        filtered_docs.append(doc)
        filtered_topics_old.append(t)

print("Documents kept:", len(filtered_docs))

# 2) Remap to new sequential labels
filtered_topics_new = [topic_mapping[t] for t in filtered_topics_old]
```

```python
from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer

# Create a new vectorizer model for the social_core_model
# with adjusted min_df since it will be applied to the small number of core topics (4).
# A min_df of 1 ensures that terms appearing in at least one topic are considered.
core_vectorizer_model = CountVectorizer(
    stop_words="english",
    ngram_range=(1,3),
    min_df=1 # Adjusted for the small number of topics
)

social_core_model = BERTopic(
    embedding_model=topic_model.embedding_model,
    vectorizer_model=core_vectorizer_model, # Use the new vectorizer
    # Removed ctfidf_model as it's better to let BERTopic build it fresh for the sub-model
    verbose=True
)

# Generate embeddings for the filtered documents
filtered_embeddings = topic_model.embedding_model.embedding_model.encode(filtered_docs)

social_core_model.fit(filtered_docs, filtered_embeddings, filtered_topics_new)

social_core_info = social_core_model.get_topic_info()
social_core_info
```

```python
fig = social_core_model.visualize_barchart(
    topics=[0,1,2],
    n_words=10,
    title="Social Core Topics (Relabeled 0–2)"
)

fig.write_html("topic_barchart_Social_core_0_3.html")
print("✓ Saved: topic_barchart_Social_core_0_3.html")
fig.show()
```

```python
def find_topic_by_keywords(keywords):
    hits = []
    info = topic_model.get_topic_info()
    for tid in info["Topic"].tolist():
        if tid == -1:
            continue
        words = topic_model.get_topic(tid)
        if not words:
            continue
        terms = [w for w,_ in words]
        score = sum(1 for k in keywords if k in terms)
        if score > 0:
            hits.append((tid, score, terms[:15]))
    hits.sort(key=lambda x: x[1], reverse=True)
    return hits

hits = find_topic_by_keywords(["factory","garment","minimum","wage","workers"])
hits[:10]
```
